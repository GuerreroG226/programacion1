Invalidación

```python
import re
class CorreoInvalidoError(Exception):
    pass
def validar_correo(correo):
    requer = r"^[a-z]+_[0-9]{1,2}_[a-z]+@universitaria\.edu\.co$"
    if not re.fullmatch(requer, correo):
        raise CorreoInvalidoError("El correo institucional es inválido.")
    return True
```

Solicitud y datos

```python
intentos_max = 3
intentos = 0
while intentos < intentos_max:
    try:
        correo = input("Ingrese su correo institucional: ")
        validar_correo(correo)
        print("Correo válido. Acceso permtido.")
        break
    except CorreoInvalidoError as error:
        intentos += 1
        print(f"Error: {error}")
        print(f"Intento {intentos} de {intentos_max}")
    except Exception as e:
        intentos += 1
        print("Error inesperado:", e)
if intentos == intentos_max:
    print("Acceso bloqueado por intentos fallidos.")
```
