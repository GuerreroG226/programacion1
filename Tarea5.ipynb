```python
Tabla_Tabaco1['Cuando_Pvez'] = Tabla_Tabaco1['Cuando_Pvez'].fillna(0)               #3
Tabla_Tabaco1['Fumado_Ult12meses']= Tabla_Tabaco1['Fumado_Ult12meses'].fillna(9)    #4
Tabla_Tabaco1['Fumado_Ult30dias'] = Tabla_Tabaco1['Fumado_Ult30dias'].fillna(9)     #5
Tabla_Tabaco1['Cuantos_diasxmes'] = Tabla_Tabaco1['Cuantos_diasxmes'].fillna(0)       #6
Tabla_Tabaco1['Fumo_Sem_Pasada'] = Tabla_Tabaco1['Fumo_Sem_Pasada'].fillna(9)       #7
Tabla_Tabaco1['Frecu_Sem_Pasada'] = Tabla_Tabaco1['Frecu_Sem_Pasada'].fillna(9)     #8
Tabla_Tabaco1['Frecuencia_diaria'] = Tabla_Tabaco1['Frecuencia_diaria'].fillna(0)     #9
Tabla_Tabaco1['Elec_Vape_ALGvez'] = Tabla_Tabaco1['Elec_Vape_ALGvez'].fillna(0)     #10
Tabla_Tabaco1['QueEdad_E_10'] = Tabla_Tabaco1['QueEdad_E_10'].fillna(0)               #11
Tabla_Tabaco1['FrecuenciaMes_E_10'] = Tabla_Tabaco1['FrecuenciaMes_E_10'].fillna(9) #12
Tabla_Tabaco1.head(20)
```

```python
Tabla_Tabaco1 = Tabla_Tabaco1.dropna()
Tabla_Tabaco1.head(20)
```

```python
Tabla_Tabaco1['Ha_consumido'] = Tabla_Tabaco1['Ha_consumido'].astype(str)
Tabla_Tabaco1['Cuando_Pvez'] = Tabla_Tabaco1['Cuando_Pvez'].astype(str)
Tabla_Tabaco1['Fumado_Ult12meses']= Tabla_Tabaco1['Fumado_Ult12meses'].astype(str)
Tabla_Tabaco1['Fumado_Ult30dias'] = Tabla_Tabaco1['Fumado_Ult30dias'].astype(str)

Tabla_Tabaco1['Fumo_Sem_Pasada'] = Tabla_Tabaco1['Fumo_Sem_Pasada'].astype(str)
Tabla_Tabaco1['Frecu_Sem_Pasada'] = Tabla_Tabaco1['Frecu_Sem_Pasada'].astype(str)


Tabla_Tabaco1['FrecuenciaMes_E_10'] = Tabla_Tabaco1['FrecuenciaMes_E_10'].astype(str)
```

```python
Tabla_Tabaco1['Cuantos_diasxmes'] = Tabla_Tabaco1['Cuantos_diasxmes'].astype(int)
Tabla_Tabaco1['Frecuencia_diaria'] = Tabla_Tabaco1['Frecuencia_diaria'].astype(int)
Tabla_Tabaco1['QueEdad_E_10'] = Tabla_Tabaco1['QueEdad_E_10'].astype(int)
```

```python
datae = pd.get_dummies(
    Tabla_Tabaco1,
    columns=['Ha_consumido', 'Cuando_Pvez', 'Fumado_Ult12meses', 'Fumado_Ult30dias', 'Fumo_Sem_Pasada', 'Frecu_Sem_Pasada', 'FrecuenciaMes_E_10']
)
datae.head(20)
```

```python
for columna in datae.columns:
    if datae[columna].dtype == bool:
        datae[columna] = datae[columna].astype(int)

display(datae.head(20))
```

```python
datae.value_counts('Elec_Vape_ALGvez')
```

```python
from sklearn.model_selection import train_test_split
train_data, test_data = train_test_split(datae, test_size=0.2, random_state=42)
```

```python
X_train = train_data.drop('Elec_Vape_ALGvez', axis=1)
y_train = train_data['Elec_Vape_ALGvez']

X_test = test_data.drop('Elec_Vape_ALGvez', axis=1)
y_test = test_data['Elec_Vape_ALGvez']
```

```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
lda = LinearDiscriminantAnalysis()
lda.fit(X_train, y_train)
```

```python
y_pred = lda.predict(X_test)
print("Predicciones:", y_pred)
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
cm = confusion_matrix(y_test, y_pred)
print(cm)
```

```python
disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot()
plt.title("Matriz de Confusión - LDA")
plt.show()
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report

modelo_rf = RandomForestClassifier(
    n_estimators=400,
    class_weight='balanced',
    random_state=42
)

modelo_rf.fit(X_train, y_train)

y_pred_rf = modelo_rf.predict(X_test)

print(classification_report(y_test, y_pred_rf))
```

```python
from sklearn.linear_model import LogisticRegression

modelo_log = LogisticRegression(
    max_iter=1000,
    class_weight='balanced'
)

modelo_log.fit(X_train, y_train)

y_pred_log = modelo_log.predict(X_test)

print(classification_report(y_test, y_pred_log))
```
