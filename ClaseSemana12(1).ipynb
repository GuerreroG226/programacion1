# Escritura de archivos en Python

```python
with open('archivo.txt', 'w') as f:
    f.write('Hola, este es un archivo de texto.\n')
    f.write('Este es el segundo renglón del archivo.\n')
```

"w" es para editar el objeto o arguemnto dado.

```python
with open('archivo.txt', 'r') as f:
  print(f.read())
```

aqui "r" es para otra funcion, por ejemplo, leerlo.

```python
from google.colab import files
files.download("archivo.txt")
```

```python
with open('archivo.txt', 'a') as f:
    f.write('\nEste es el tercer renglón del archivo.')
```

```python
with open('archivo.txt', 'r') as f:
  print(f.read())
```

```python
with open('archivo.txt', 'r') as f:
    lines = f.readlines()

# Eliminar la última línea si el archivo no está vacío
if lines:
    lines = lines[:-1]

with open('archivo.txt', 'w') as f:
    f.writelines(lines)
```

Ahora, si leemos el archivo de nuevo, la última línea debería haber sido eliminada.
