*Instrucción 1:*

```python
A = {1, 2, 3, 4}
B = {1, 2, 5, 6}
```

```python
A.difference(B)
```

El resultado de esta operación será un conjunto con los elemento diferentes entre ellos. Como "1" y "2" están en ambos, la salida será "3" y"4". No saldrá "5" y "6" porque esta acción reace en A.

```python
A.symmetric_difference(B)
```

Aquí la acción recae en ambos conjuntos, o sea, la salida será de "3", "4", "5" y "6".
