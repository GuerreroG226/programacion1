**PRUEBA A**

```python
import numpy as np
import pandas as pd

np.random.seed(1)

datoss = np.random.normal(loc=7.2, scale=0.4, size= 5 * int(W[-1]))

fg = pd.DataFrame({'Datos': datoss})

fg
```

**PRUEBA B**

```python
np.random.seed(1)

datoss_gamma = np.random.gamma(shape=306, scale=0.0229, size=5 * int(W[-1]))

fg_gamma = pd.DataFrame({'Datos Gamma': datoss_gamma})

fg_gamma
```

DataFrame

```python
am = pd.DataFrame({'Datos': datoss, 'Datos Gamma': datoss_gamma})
am
```

```python
print(am['Datos Gamma'].describe())

print(am['Datos'].describe())
```

Analisis:


```python
print('Media               :', am['Datos'].mean())

print('Mediana             :', am['Datos'].median())

print('Minimo              :', am['Datos'].min())

print('Maximo              :', am['Datos'].max())

print('Desviación estándar :', am['Datos'].std())

print('Cuantil 25%         :', am['Datos'].quantile(0.25))

print('Cuantil 75%         :', am['Datos'].quantile(0.75))
```

prueba de shapiro

```python
from scipy.stats import shapiro
```

```python
shapiro(am['Datos'])
```

```python
shapiro(am['Datos Gamma'])
```

Podemos ver que ambos Datos alcanzan un valor p mayor que 0.05.
entonces sí se puede admitir este valor como un 95% acertado respecto a la hipótesis del análisis.

**prueba t student**

```python
import numpy as np
import pandas as pd
import scipy.stats as stats
```

```python
t_rel, p_rel = stats.ttest_rel(am['Datos'], am['Datos Gamma'])
print("t-statistic:", t_rel)
print("p-value:", p_rel)
```

El valor p es menor a 0.05, entonces afirmamos con un 95% de confianza que las medias no son iguales

**Kendall**

```python
t_sta, p_valu = stats.kendalltau(am['Datos'], am['Datos Gamma'])
print("t-statistic:", t_sta)
print("p-value:", p_valu)
```

podemos decir con 95% de confianza que ambas encuestas no estuvieron relacionadas, dado que el valor p es menor a 0.05

**Spearman**

```python
s_pea, p_value = stats.spearmanr(am['Datos'], am['Datos Gamma'])
print("t-statistic:", s_pea)
print("p-value:", p_value)
```

podemos decir con un 95% de confianza que los datos no estuvieron correlacionas dado que el valor p es menor a 0.05

```python
%%writefile comparacion/pruebat.py

import numpy as np
import pandas as pd
import scipy.stats as stats

def ttest_rel(data1, data2):
    t_rel, p_rel = stats.ttest_rel(data1, data2)
    print("t-statistic (prueba t Student):", t_rel)
    print("p-value (prueba t Student):", p_rel)
```

```python
%%writefile comparacion/cor.py


import numpy as np
import pandas as pd
import scipy.stats as stats

def kendall(data1, data2):
    t_sta, p_valu = stats.kendalltau(data1, data2)
    print("t-statistic (Kendall):", t_sta)
    print("p-value (Kendall):", p_valu)


def spearman(data1, data2):
  t_sta, p_value = stats.spearmanr(data1, data2)
  print("t-statistic (Spearman):", t_sta)
  print("p-value (Spearman):", p_value)
```
